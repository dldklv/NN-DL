```python
import ipywidgets as widgets
from IPython.display import HTML, display


class UltimateSnakeGame:

    def __init__(self):
        self.game_html = """
        <div id="gameContainer" style="text-align: center; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; padding: 20px; border-radius: 12px; max-width: 750px; margin: 0 auto; transition: all 0.3s ease;">
            <h2 style="margin-top: 0;">🐍 Ultimate Snake: Bonuslar & To'siqlar</h2>

            <!-- Interfeys / Mavzu tanlash -->
            <div style="margin-bottom: 15px;">
                <label for="themeSelect" style="font-weight: bold; margin-right: 8px;">🎨 Interfeys Mavzusi:</label>
                <select id="themeSelect" onchange="changeTheme(this.value)" style="padding: 6px 12px; font-size: 14px; border-radius: 6px; cursor: pointer;">
                    <option value="classic">🟢 Klassik (Yashil)</option>
                    <option value="neon">🌌 Kiberpunk (Neon)</option>
                    <option value="retro">📜 Retro (GameBoy)</option>
                    <option value="dark">🌙 Tungi (Dark Modern)</option>
                    <option value="beach">🏖️ Plaj (Yozgi)</option>
                </select>
            </div>

            <!-- Hisob, Level va Holat -->
            <div style="display: flex; justify-content: space-around; font-size: 16px; font-weight: bold; margin-bottom: 10px; background: rgba(0,0,0,0.05); padding: 8px; border-radius: 6px;">
                <div>Bosqich (Level): <span id="level" style="color: #2196F3;">1</span></div>
                <div>Ochko: <span id="score" style="color: #4CAF50;">0</span></div>
                <div>Rekord: <span id="highScore" style="color: #FF9800;">0</span></div>
            </div>

            <!-- Bonuslar haqida ma'lumot -->
            <div style="font-size: 12px; margin-bottom: 8px; color: #666;">
                🔴 Oddiy (+10) | 🌟 Oltin (+30) | ❄️ Muz (+15 & Sekinlashtirish) | 🟥 To'siq
            </div>

            <div style="display: flex; justify-content: center; gap: 20px; flex-wrap: wrap;">
                <!-- O'yin maydoni -->
                <div style="position: relative;">
                    <canvas id="snakeCanvas" width="360" height="360" style="border: 4px solid; border-radius: 8px; box-shadow: 0 4px 10px rgba(0,0,0,0.3);"></canvas>

                    <!-- Game Over Overlay -->
                    <div id="gameOverOverlay" style="display: none; position: absolute; top: 0; left: 0; width: 100%; height: 100%; background: rgba(0,0,0,0.88); color: white; border-radius: 8px; flex-direction: column; justify-content: center; align-items: center; padding: 20px; box-sizing: border-box;">
                        <h2 style="color: #FF5252; margin-bottom: 5px;">O'YIN TUGADI!</h2>
                        <p style="font-size: 16px; margin: 5px 0;">Yakuniy Ochko: <span id="finalScore" style="font-weight:bold; color: #4CAF50;">0</span></p>
                        <p style="font-size: 14px; margin: 2px 0; color: #2196F3;">Erishilgan Level: <span id="finalLevel">1</span></p>

                        <div id="nameInputBox" style="margin: 10px 0; display: none;">
                            <p style="margin: 5px; color: #FFD700; font-size: 13px;">🎉 Tabriklaymiz! Top-10 ga kirdingiz!</p>
                            <input type="text" id="playerName" placeholder="Ismingiz" style="padding: 6px; font-size: 14px; border-radius: 4px; border: none; text-align: center; width: 120px;" maxlength="12">
                            <button onclick="saveScore()" style="padding: 6px 12px; background: #2196F3; color: white; border: none; border-radius: 4px; cursor: pointer;">Saqlash</button>
                        </div>

                        <button onclick="startGame()" style="margin-top: 12px; padding: 10px 20px; font-size: 15px; background-color: #4CAF50; color: white; border: none; border-radius: 5px; cursor: pointer; font-weight: bold;">🔄 Qaytadan Boshlash</button>
                    </div>
                </div>

                <!-- Top 10 Rekordlar Jadvali -->
                <div id="leaderboardBox" style="width: 200px; padding: 10px; border-radius: 8px; text-align: left; box-sizing: border-box;">
                    <h3 style="margin-top: 0; text-align: center; font-size: 15px; border-bottom: 2px solid; padding-bottom: 5px;">🏆 TOP-10 Rekordlar</h3>
                    <ol id="leaderboardList" style="padding-left: 22px; font-size: 12px; margin: 0; line-height: 1.6;">
                    </ol>
                </div>
            </div>

            <!-- Boshqaruv tugmalari -->
            <div style="margin-top: 15px;">
                <button onclick="startGame()" style="padding: 8px 25px; font-size: 14px; background-color: #4CAF50; color: white; border: none; border-radius: 5px; cursor: pointer; font-weight: bold;">Yangi O'yin</button>
            </div>

            <div style="margin-top: 10px;">
                <div>
                    <button onclick="changeDir('UP')" style="padding: 6px 12px; margin: 2px; cursor: pointer;">⬆️</button>
                </div>
                <div>
                    <button onclick="changeDir('LEFT')" style="padding: 6px 12px; margin: 2px; cursor: pointer;">⬅️</button>
                    <button onclick="changeDir('DOWN')" style="padding: 6px 12px; margin: 2px; cursor: pointer;">⬇️</button>
                    <button onclick="changeDir('RIGHT')" style="padding: 6px 12px; margin: 2px; cursor: pointer;">➡️</button>
                </div>
            </div>
        </div>

        <script>
            var canvas = document.getElementById("snakeCanvas");
            var ctx = canvas.getContext("2d");
            var scoreElement = document.getElementById("score");
            var levelElement = document.getElementById("level");
            var highScoreElement = document.getElementById("highScore");
            var gameOverOverlay = document.getElementById("gameOverOverlay");
            var nameInputBox = document.getElementById("nameInputBox");

            var grid = 18;
            var score = 0;
            var level = 1;
            var speed = 100;
            var gameTimeout;
            var currentTheme = 'classic';

            var themes = {
                classic: {
                    containerBg: "#f4f4f4", canvasBg: "#111111", canvasBorder: "#4CAF50",
                    snake: "#4CAF50", apple: "#FF5252", obstacle: "#D32F2F", text: "#333333", leaderboardBg: "#ffffff"
                },
                neon: {
                    containerBg: "#0f0826", canvasBg: "#05010d", canvasBorder: "#00f5d4",
                    snake: "#00f5d4", apple: "#f72585", obstacle: "#7209b7", text: "#00f5d4", leaderboardBg: "#1a0c36"
                },
                retro: {
                    containerBg: "#9bbc0f", canvasBg: "#8bac0f", canvasBorder: "#0f380f",
                    snake: "#0f380f", apple: "#306230", obstacle: "#000000", text: "#0f380f", leaderboardBg: "#8bac0f"
                },
                dark: {
                    containerBg: "#212121", canvasBg: "#121212", canvasBorder: "#BB86FC",
                    snake: "#BB86FC", apple: "#CF6679", obstacle: "#FFB74D", text: "#E1E1E1", leaderboardBg: "#2c2c2c"
                },
                beach: {
                    containerBg: "#fff8e1", canvasBg: "#ffecb3", canvasBorder: "#ff9800",
                    snake: "#00acc1", apple: "#e91e63", obstacle: "#795548", text: "#006064", leaderboardBg: "#ffe0b2"
                }
            };

            var snake = { x: 144, y: 144, dx: grid, dy: 0, cells: [], maxCells: 4 };
            var apple = { x: 216, y: 216 };
            var bonusApple = { x: -1, y: -1, type: null, timer: null };
            var obstacles = [];

            function getRandomInt(min, max) {
                return Math.floor(Math.random() * (max - min)) + min;
            }

            function generateFreePos() {
                while (true) {
                    var rx = getRandomInt(0, 20) * grid;
                    var ry = getRandomInt(0, 20) * grid;

                    var onSnake = snake.cells.some(c => c.x === rx && c.y === ry);
                    var onObs = obstacles.some(o => o.x === rx && o.y === ry);
                    var onApple = (apple.x === rx && apple.y === ry);

                    if (!onSnake && !onObs && !onApple) {
                        return { x: rx, y: ry };
                    }
                }
            }

            function updateObstacles() {
                obstacles = [];
                // Har bir Level uchun maxsus to'siqlar joylashuvi
                if (level >= 2) {
                    // Markaziy to'siqlar
                    obstacles.push({x: 5*grid, y: 10*grid}, {x: 6*grid, y: 10*grid}, {x: 7*grid, y: 10*grid});
                    obstacles.push({x: 12*grid, y: 10*grid}, {x: 13*grid, y: 10*grid}, {x: 14*grid, y: 10*grid});
                }
                if (level >= 3) {
                    // Burchak to'siqlar
                    obstacles.push({x: 3*grid, y: 3*grid}, {x: 3*grid, y: 4*grid}, {x: 4*grid, y: 3*grid});
                    obstacles.push({x: 16*grid, y: 16*grid}, {x: 16*grid, y: 15*grid}, {x: 15*grid, y: 16*grid});
                }
                if (level >= 4) {
                    // Qo'shimcha ustunlar
                    for(var i=6; i<=13; i+=2) {
                        obstacles.push({x: i*grid, y: 5*grid});
                        obstacles.push({x: i*grid, y: 14*grid});
                    }
                }
            }

            function spawnBonusApple() {
                if (Math.random() < 0.4 && bonusApple.x === -1) { // 40% imkoniyat bilan paydo bo'ladi
                    var pos = generateFreePos();
                    bonusApple.x = pos.x;
                    bonusApple.y = pos.y;
                    bonusApple.type = Math.random() < 0.5 ? 'gold' : 'freeze';

                    // 5 soniyadan keyin yo'qoladi
                    clearTimeout(bonusApple.timer);
                    bonusApple.timer = setTimeout(removeBonusApple, 5000);
                }
            }

            function removeBonusApple() {
                bonusApple.x = -1;
                bonusApple.y = -1;
                bonusApple.type = null;
            }

            function changeTheme(themeKey) {
                currentTheme = themeKey;
                var t = themes[themeKey];

                document.getElementById("gameContainer").style.backgroundColor = t.containerBg;
                document.getElementById("gameContainer").style.color = t.text;
                canvas.style.backgroundColor = t.canvasBg;
                canvas.style.borderColor = t.canvasBorder;

                var lbBox = document.getElementById("leaderboardBox");
                lbBox.style.backgroundColor = t.leaderboardBg;
                lbBox.style.border = "1px solid " + t.canvasBorder;

                drawGame();
            }

            function loadLeaderboard() {
                var leaderboard = JSON.parse(localStorage.getItem("snake_top10_v2") || "[]");
                var listHtml = "";
                leaderboard.forEach(function(item) {
                    listHtml += "<li><b>" + item.name + "</b>: " + item.score + " (Lvl " + item.level + ")</li>";
                });
                document.getElementById("leaderboardList").innerHTML = listHtml || "<i>Hali rekordlar yo'q</i>";

                var topScore = leaderboard.length > 0 ? leaderboard[0].score : 0;
                highScoreElement.innerHTML = topScore;
            }

            function saveScore() {
                var pName = document.getElementById("playerName").value.trim() || "O'yinchi";
                var leaderboard = JSON.parse(localStorage.getItem("snake_top10_v2") || "[]");

                leaderboard.push({ name: pName, score: score, level: level });
                leaderboard.sort((a, b) => b.score - a.score);
                leaderboard = leaderboard.slice(0, 10);

                localStorage.setItem("snake_top10_v2", JSON.stringify(leaderboard));
                nameInputBox.style.display = "none";
                loadLeaderboard();
            }

            function checkTop10() {
                var leaderboard = JSON.parse(localStorage.getItem("snake_top10_v2") || "[]");
                if (score > 0 && (leaderboard.length < 10 || score > leaderboard[leaderboard.length - 1].score)) {
                    nameInputBox.style.display = "block";
                } else {
                    nameInputBox.style.display = "none";
                }
            }

            function startGame() {
                snake.x = 144;
                snake.y = 144;
                snake.cells = [];
                snake.maxCells = 4;
                snake.dx = grid;
                snake.dy = 0;
                score = 0;
                level = 1;
                speed = 100;

                scoreElement.innerHTML = score;
                levelElement.innerHTML = level;

                removeBonusApple();
                updateObstacles();

                var aPos = generateFreePos();
                apple.x = aPos.x;
                apple.y = aPos.y;

                gameOverOverlay.style.display = "none";

                clearTimeout(gameTimeout);
                gameLoop();
            }

            function changeDir(dir) {
                if (dir === 'LEFT' && snake.dx === 0) { snake.dx = -grid; snake.dy = 0; }
                else if (dir === 'UP' && snake.dy === 0) { snake.dy = -grid; snake.dx = 0; }
                else if (dir === 'RIGHT' && snake.dx === 0) { snake.dx = grid; snake.dy = 0; }
                else if (dir === 'DOWN' && snake.dy === 0) { snake.dy = grid; snake.dx = 0; }
            }

            document.addEventListener('keydown', function(e) {
                if (e.which === 37) changeDir('LEFT');
                else if (e.which === 38) changeDir('UP');
                else if (e.which === 39) changeDir('RIGHT');
                else if (e.which === 40) changeDir('DOWN');
            });

            function endGame() {
                clearTimeout(gameTimeout);
                document.getElementById("finalScore").innerHTML = score;
                document.getElementById("finalLevel").innerHTML = level;
                checkTop10();
                gameOverOverlay.style.display = "flex";
            }

            function drawGame() {
                var t = themes[currentTheme];
                ctx.clearRect(0, 0, canvas.width, canvas.height);

                // To'siqlarni chizish
                ctx.fillStyle = t.obstacle;
                obstacles.forEach(function(obs) {
                    ctx.fillRect(obs.x, obs.y, grid - 1, grid - 1);
                });

                // Oddiy Olmani chizish
                ctx.fillStyle = t.apple;
                ctx.beginPath();
                ctx.arc(apple.x + grid/2, apple.y + grid/2, grid/2 - 2, 0, 2 * Math.PI);
                ctx.fill();

                // Bonus Olmani chizish
                if (bonusApple.x !== -1) {
                    ctx.fillStyle = bonusApple.type === 'gold' ? '#FFD700' : '#00BFFF';
                    ctx.beginPath();
                    ctx.arc(bonusApple.x + grid/2, bonusApple.y + grid/2, grid/2 - 1, 0, 2 * Math.PI);
                    ctx.fill();

                    // Ichki effekt
                    ctx.fillStyle = '#FFFFFF';
                    ctx.beginPath();
                    ctx.arc(bonusApple.x + grid/3, bonusApple.y + grid/3, 2, 0, 2 * Math.PI);
                    ctx.fill();
                }

                // Ilonchani chizish
                ctx.fillStyle = t.snake;
                snake.cells.forEach(function(cell, idx) {
                    ctx.fillRect(cell.x, cell.y, grid - 1, grid - 1);
                });
            }

            function gameLoop() {
                snake.x += snake.dx;
                snake.y += snake.dy;

                // Devorga urilish
                if (snake.x < 0 || snake.x >= canvas.width || snake.y < 0 || snake.y >= canvas.height) {
                    endGame();
                    return;
                }

                // To'siqlarga urilish
                for (var i = 0; i < obstacles.length; i++) {
                    if (snake.x === obstacles[i].x && snake.y === obstacles[i].y) {
                        endGame();
                        return;
                    }
                }

                snake.cells.unshift({x: snake.x, y: snake.y});
                if (snake.cells.length > snake.maxCells) snake.cells.pop();

                // Oddiy olmani yeyish
                if (snake.x === apple.x && snake.y === apple.y) {
                    snake.maxCells++;
                    score += 10;

                    // Level oshirish
                    var newLevel = Math.floor(score / 50) + 1;
                    if (newLevel !== level) {
                        level = newLevel;
                        levelElement.innerHTML = level;
                        speed = Math.max(40, 100 - (level - 1) * 12); // Har levelda tezlashadi
                        updateObstacles();
                    }

                    scoreElement.innerHTML = score;
                    var aPos = generateFreePos();
                    apple.x = aPos.x;
                    apple.y = aPos.y;

                    spawnBonusApple();
                }

                // Bonus olmani yeyish
                if (snake.x === bonusApple.x && snake.y === bonusApple.y) {
                    if (bonusApple.type === 'gold') {
                        score += 30;
                    } else if (bonusApple.type === 'freeze') {
                        score += 15;
                        speed += 30; // Vaqtincha sekinlashtiradi
                    }
                    scoreElement.innerHTML = score;
                    removeBonusApple();
                }

                // O'ziga urilish
                for (var i = 1; i < snake.cells.length; i++) {
                    if (snake.x === snake.cells[i].x && snake.y === snake.cells[i].y) {
                        endGame();
                        return;
                    }
                }

                drawGame();
                gameTimeout = setTimeout(gameLoop, speed);
            }

            // Boshlang'ich yuklash
            changeTheme('classic');
            loadLeaderboard();
        </script>
        """

    def show(self):
        display(HTML(self.game_html))


# O'yinni ishga tushirish
game = UltimateSnakeGame()
game.show()
```
